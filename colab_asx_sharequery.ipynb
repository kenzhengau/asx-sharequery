```python
# 3B) Run the browser chat window (DevUI) behind a public tunnel
# This is the easiest way to expose the app from Colab.
# 1) install ngrok
# 2) start the app on a local port
# 3) use a tunnel to publish it

!python -m pip install --quiet pyngrok
from pyngrok import ngrok

!cd /content/asx-sharequery && nohup python -m asx_sharequery.webui --host 0.0.0.0 --port 8000 >/tmp/asx_sharequery.log 2>&1 &
print('Starting tunnel...')
public_url = ngrok.connect(8000, 'http')
print(public_url)
```
